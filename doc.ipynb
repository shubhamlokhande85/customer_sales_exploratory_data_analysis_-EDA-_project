The project follows this workflow:

##### Data Inspection → Data Cleaning → Data Transformation → Analysis → Visualization → Business Insights → Recommendations

### 4. Business Problems and What I Found


##### Problem 1: How are purchase amounts distributed across gender categories?

What I found:

Male customers contributed approximately 46.2% of recorded purchase amounts.

Female customers contributed approximately 45.1%.

The Unknown category contributed approximately 8.6%.

Business Insight: The male and female categories contributed similar proportions of the recorded purchase amounts. The Unknown category represents an opportunity to improve demographic data completeness.


##### Problem 2: Which age groups contribute the highest purchase amounts?

What I found:

The 45–60 age group had the highest total recorded purchase amount in the displayed chart.

The 0–18 age group had the lowest total.

The middle-aged groups contributed more than the youngest group.

Business Insight: The 30–60 age groups may be useful segments for further investigation. However, total purchase amount alone does not show whether these customers spend more individually or simply represent more records.


##### Problem 3: How do purchase amounts change month by month?

What I found:

Recorded purchase amounts showed a sharp increase in March 2025.

Another prominent increase occurred in July 2025.

Amounts declined after the July peak through September.

October 2025 showed a substantial decrease compared with September.

Business Insight: These fluctuations identify months that need further investigation. The reasons could include unusual records, incomplete data, or genuine purchasing patterns.


##### Problem 4: Is customer age related to purchase amount?

What I found:

The correlation between Age and Purchase_Amount was approximately 0.02, indicating almost no linear relationship in the analyzed data.

Business Insight: Age alone did not show a meaningful linear association with purchase amount. Other factors may be more useful for understanding customer spending.


##### Problem 5: Is customer feedback related to purchase amount?

What I found:

The correlation between Feedback_Score and Purchase_Amount was close to zero.

Business Insight: The analysis did not show a meaningful linear relationship between feedback score and purchase amount. Further analysis is needed before concluding whether customer satisfaction influences purchasing behavior.

##### Problem 6: Are there data-quality issues that could affect business decisions?

What I found:

Invalid age entries included Unknown, -10, and 250.

Gender values contained inconsistent formatting.

The Last_Purchase_Date column contained 1,009 missing values.

Four records had negative purchase amounts.

The maximum purchase amount was 9,999,999, which requires verification.

Business Insight: Data-quality issues can affect demographic comparisons and trend analysis. Validating unusual values and documenting cleaning assumptions are important for reliable business reporting.

### 7. Business Recommendations

Improve Data Quality: Introduce validation rules for age, dates, gender categories, and purchase amounts.

Investigate Unusual Values: Verify negative amounts and the unusually large purchase amount before using them for business decisions.

Improve Customer Segmentation: Compare customer counts and average purchase amounts across age groups and gender categories.

Monitor Monthly Trends: Investigate the March and July peaks and October decline, checking whether reporting periods are complete.

Improve Customer Insights: Explore feedback patterns, repeat purchases, customer retention, and geographic performance when suitable data is available.

Improve Correlation Analysis: Exclude phone numbers from numerical correlation analysis because they are identifiers rather than meaningful measurements.

Track KPIs Regularly: Monitor total customers, total purchase amount, average purchase amount, and average feedback score over time to identify changes in customer behavior and business performance.
